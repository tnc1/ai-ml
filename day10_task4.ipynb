```python
"""
  TASK 4 — STUDENT FEEDBACK SENTIMENT ANALYZER
    The university receives feedback comments from students. The administration wants AI to identify whether comments are:
      ● Positive
      ● Negative
      ● Neutral

  Sample Feedback Feedback

  “Faculty support is excellent” “Assignments are stressful” “Labs are average”

  Tasks
    1. Tokenize the feedback sentence.
    2. Remove stopwords.
    3. Use SentimentIntensityAnalyzer.
    4. Display sentiment scores.
    5. Display final sentiment category.

"""

# STEP 1 — Import Libraries

import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.sentiment import SentimentIntensityAnalyzer


# STEP 2 — Download Required NLTK Data

nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('vader_lexicon')


# STEP 3 — Load spaCy Model

!python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")


# STEP 4 — Input Review

text = input("Enter feedback from students to analyze sentiment: ")

# STEP 5 — Tokenization

tokens = word_tokenize(text)

print("Tokens:\n")

print(tokens)


# STEP 6 — Stopwords Removal

stop_words = set(stopwords.words('english'))

filtered_words = []

for word in tokens:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print("\nAfter Stopwords Removal:\n")

print(filtered_words)


# STEP 7 — Lemmatization

lemmatizer = WordNetLemmatizer()

lemmatized_words = []

for word in filtered_words:
    lemmatized_words.append(
        lemmatizer.lemmatize(word)
    )

print("\nLemmatized Words:\n")

print(lemmatized_words)


# STEP 8 — Sentiment Analysis

sia = SentimentIntensityAnalyzer()

sentiment = sia.polarity_scores(text)

if sentiment['compound'] >= 0.05:
    prediction = "positive"
elif sentiment['compound'] <= -0.05:
    prediction = "negative"
else:
    prediction = "neutral"

print(f"\nSentiment score is {prediction}")

```
