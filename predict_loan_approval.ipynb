```python
'''
Scenario: Predicting Loan Approval Based on Income and Credit Score

Context:

  A bank wants to decide whether to approve or reject a loan application based on two factors:

    - Applicant’s Annual Income
    - Applicant’s Credit Score

  They believe there are clear threshold-based rules (e.g., higher income and good credit → approval), so they use a Decision Tree Classifier.

  Accuracy score, confusion matrix.
'''

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix

# Step 1: Load dataset
data = pd.DataFrame({
    'Annual_Income': [2.5,3.0,4.5,5.0,6.0,3.5,7.0,8.0,2.0,4.0],
    'Credit_Score': [580,600,650,700,720,590,750,780,560,640],
    'Loan_Status': ['Reject','Reject','Approve','Approve','Approve','Reject','Approve','Approve','Reject','Approve']
})

# Step 2: Visualize
plt.scatter(data['Annual_Income'], data['Loan_Status'])
plt.xlabel("Annual Income")
plt.ylabel("Loan Status")
plt.title("Annual Income vs Credit Score")
plt.show()

# Step 3: Train Decision Tree
X = data[['Annual_Income','Credit_Score']]
y = data['Loan_Status']

model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X, y)

# Step 4: Predict
prediction = model.predict(
    pd.DataFrame([[5.0, 600]], columns=X.columns)
)
print("Predicted Result for 5.0 annual income and 600 credit score:", prediction[0])

# Step 5: Evaluate
y_pred = model.predict(X)
print("Accuracy:", accuracy_score(y, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y, y_pred))

# Optional: Visualize the Decision Tree
plt.figure(figsize=(8,6))
plot_tree(model, feature_names=['Annual_Income'], class_names=model.classes_, filled=True)
plt.show()
```
